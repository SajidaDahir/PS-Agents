## Practical Exercise 3 Extension: Uniform Cost Search (UCS)

```python
import heapq

#Defining the Kenyan Highway Network
graph = {
    'Nairobi':[('Naivasha', 88),('Narok', 142)],
    'Naivasha':[('Nakuru', 71)],
    'Nakuru':[('Molo', 47),('Eldoret', 156)],
    'Molo':[('Kericho', 60)],
    'Kericho':[('Kisumu', 83)],
    'Narok':[('Bomet', 86)],
    'Bomet':[('Kisii', 68)],
    'Kisii':[('Kisumu', 114)],
    'Eldoret':[('Kapsabet', 48)],
    'Kapsabet':[('Kericho', 74)],
    'Kisumu':[] # This is the goal state
}

#Implementing a Uniform Cost Search
def uniform_cost_search(graph, start_node, goal_node):
  frontier = []
  heapq.heappush(frontier, (0, start_node, [start_node]))

  explored = set()
  step = 1

  print(f"Starting Uniform Cost Search: {start_node} {goal_node} \n")
  while frontier:
    current_cost, current_node, path = heapq.heappop(frontier)
    print(f"Step {step}: Expanding Node -> '{current_node}' (Cost: {current_cost} km)")
    step +=1

    #Testing the goal
    if current_node == goal_node:
      print(f"Goal Reached! The optimal path to {goal_node} has been found.")
      return path, current_cost

    if current_node not in explored:
      explored.add(current_node)

      for neighbour, weight in graph.get(current_node, []):
        if neighbour not in explored:
          new_cost = current_cost + weight
          new_path = path + [neighbour]
          heapq.heappush(frontier,(new_cost,neighbour, new_path))
          print(f"Frontier Pushed: {neighbour} (Total: {new_cost} km)")
    print()

  return None, float('inf')

#Executing the search now
optimal_path, path_cost = uniform_cost_search(graph, 'Nairobi', 'Kisumu')

#Printing the results
print("Final Results")
print(f"Optimal Path: {'->'.join(optimal_path)}")
print(f"Total Distance: {path_cost} km")
```
